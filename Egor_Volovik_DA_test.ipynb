````python
PRIORITIES = [
    (['Код товара'],            ['Продукт'],                  True),
    (['Бренд', 'Категория3'],   ['Бренд', 'Категория'],       False),
    (['Бренд', 'Категория2'],   ['Бренд', 'Категория'],       False),
    (['Бренд', 'Категория1'],   ['Бренд', 'Категория'],       False),
    (['Бренд'],                 ['Бренд'],                    True ),
    (['Сегмент', 'Категория3'], ['Сегмент рынка', 'Категория'], False),
    (['Сегмент', 'Категория2'], ['Сегмент рынка', 'Категория'], False),
    (['Сегмент', 'Категория1'], ['Сегмент рынка', 'Категория'], False),
    (['Сегмент'],               ['Сегмент рынка'],            True ),
]

def apply_markup_priority(t1: pd.DataFrame, overprie: pd.DataFrame) -> pd.DataFrame:
    result = t1.copy()
    result['Наценка'] = np.nan

    for i, (left_on, right_on, cat_isna) in enumerate(PRIORITIES):
        d2 = overprie[overprie['Категория'].isna()] if cat_isna else overprie

        d2 = (d2.dropna(subset=right_on + ['Наценка'])
                .loc[:, right_on + ['Наценка']]
                .rename(columns=dict(zip(right_on, left_on))))


        eligible = result['Наценка'].isna()
        if not eligible.any() or d2.empty:
            continue

        lookup = result.loc[eligible, left_on].merge(d2, on=left_on, how='left')
        result.loc[eligible, 'Наценка'] = lookup['Наценка'].to_numpy()

    return result
````

Для каждого SKU считаю ads для каждого уровня цены (5-7). По этим точкам строю регрессию в логарифмах log(продажи) - log(относительная цена); её наклон - это и есть эластичность E, то есть на сколько процентов меняются продажи при изменении цены на 1%. Из предположения о постоянной эластичности вывожу скидку, при которой ads удваиваются относительно уровня без скидки.

````python
from sklearn.linear_model import LinearRegression

t2['r'] = t2['Цена с учётом скидки'] / t2['Розничная цена']

levels = (
    t2.groupby(['Код товара', 'r'], as_index=False)
      .agg(ads=('Количество продано', 'mean'))
)

rows = []
for sku, g in levels.groupby('Код товара'):

    # Ищу эластичность
    E = LinearRegression().fit(
        np.log(g[['r']]), np.log(g['ads'])
    ).coef_[0]

    # скидка, при продажах x2
    # вывод: ADS * (1 - DISCOUNT)^(E) = 2 * ADS
    # вывод: 1 - DISCOUNT = 2^(1/E)
    # вывод: DISCOUNT = 1 - 2^(1/E)
    d = round(100 * (1 - 2 ** (1 / E)), 1)

    rows.append({'Код товара': sku, 'E': round(E, 2),
                 'скидка_%': d})

pd.DataFrame(rows)
````

````python
brand = (t3.groupby(['Бренд','Неделя'])
    .agg(Выручка=('Выручка','sum'), Маржа_млн=('Маржа_руб','sum'))
    .reset_index())
brand['Маржинальность'] = brand['Маржа_млн'] / brand['Выручка']

piv = brand.pivot(index='Бренд', columns='Неделя',
                  values=['Выручка','Маржинальность','Маржа_млн'])

brand_delta = pd.DataFrame({
    'Выручка_14_млн': piv[('Выручка',14)],
    'Выручка_15_млн': piv[('Выручка',15)],
    'ΔВыручка_млн':   piv[('Выручка',15)] - piv[('Выручка',14)],
    'ΔВыручка_%':     (piv[('Выручка',15)]/piv[('Выручка',14)] - 1)*100,
    'Марж_14_%':      piv[('Маржинальность',14)]*100,
    'Марж_15_%':      piv[('Маржинальность',15)]*100,
    'ΔМарж_п.п.':     (piv[('Маржинальность',15)] - piv[('Маржинальность',14)])*100,
    'ΔМаржа_млн':     piv[('Маржа_млн',15)] - piv[('Маржа_млн',14)],
}).sort_values('ΔМаржа_млн')

brand_delta.round(2)
````

````python
df = brand.pivot(index='Бренд', columns='Неделя', values=['Выручка','Маржинальность'])
df.columns = ['_'.join(map(str,c)) for c in df.columns]

rev14, rev15 = df['Выручка_14'].sum(), df['Выручка_15'].sum()
df['Доля_14'] = df['Выручка_14']/rev14
df['Доля_15'] = df['Выручка_15']/rev15

df['Mix_effect']  = (df['Доля_15'] - df['Доля_14']) * df['Маржинальность_14']
df['Margin_effect'] =  df['Доля_15'] * (df['Маржинальность_15'] - df['Маржинальность_14'])
df['Вклад_итого'] = df['Mix_effect'] + df['Margin_effect']

print(f"Δ общей маржинальности: {(df['Вклад_итого'].sum()*100):.2f} п.п.")
print(f"  · эффект структуры (mix): {(df['Mix_effect'].sum()*100):.2f} п.п.")
print(f"  · эффект ставок (rate):   {(df['Margin_effect'].sum()*100):.2f} п.п.\n")

print("Вклад брендов в изменение общей маржи (п.п.):")
df[['Mix_effect','Margin_effect','Вклад_итого']].mul(100).round(2).sort_values('Вклад_итого', ascending=False)
````

````python
Написать SQL-запросы на основе описанных таблиц в Clickhouse													
Рассчитать WAU скользящим окном в неделю с шагом в 1 день, при этом текущая дата должна включаться в расчет.													
WAU - кол-во активных пользователей сайта.													
Например, для даты 19.12.2023 нужно рассчитать WAU за период 13.12.2023-19.12.2023,													
для  20.12.2023  - 14.12.2023-20.12.2023													
для  21.12.2023  - 15.12.2023-21.12.2023													
и т.д.													
````

Первый способ - менее оптимальный. 

для каждой даты джоиню пользователей, которые были активны между этой датой и датой - 6

ON t2.event_date BETWEEN t1.event_date - INTERVAL 6 DAY
                         AND t1.event_date

```sql
SELECT
    t1.event_date AS report_date,
    uniqExact(t2.user_id) AS rolling_wau
FROM (
    SELECT DISTINCT toDate(timestamp) AS event_date
    FROM dwh.site_activity
) AS t1
INNER JOIN (
    SELECT
        toDate(timestamp) AS event_date,
        user_id
    FROM dwh.site_activity
    WHERE toDate(timestamp) <= today()
) AS t2
    ON t2.event_date BETWEEN t1.event_date - INTERVAL 6 DAY
                         AND t1.event_date
GROUP BY t1.event_date
ORDER BY report_date;
```

Второй способ - оптимальный. 

для каждой даты создаю хэш-множества через state функцию. После этого оптимально считаю уникальных пользователей оконкой


```sql
WITH daily_states AS (
    SELECT
        toDate(timestamp) AS event_date,
        uniqExactState(user_id) AS state
    FROM dwh.site_activity
    WHERE toDate(timestamp) <= today()
    GROUP BY event_date
)
SELECT
    event_date AS report_date,
    uniqExactMerge(state) OVER (
        ORDER BY event_date
        ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
    ) AS rolling_wau
FROM daily_states
ORDER BY report_date;
```
